**B.CARI INVERS**

menggunakan rumus Adjoin:

$$(\operatorname{adj} A)_{ij} = (-1)^{i+j} M_{ji}$$

dan Invers :

$$A^{-1} = \frac{1}{\det A} \operatorname{adj} A.$$

soal 4 :

Karna soalnya sama seperti tugas A, maka diketahui determinan dari soal 1 adalah -23

menghitung adjoin :

$C_{11} = (-1)^{1+1} (4) = 4$

$C_{12} = (-1)^{1+2} (1) = -1$

$C_{21} = (-1)^{2+1} (-5) = 5$

$C_{22} = (-1)^{2+2} (-7) = -7$

Maka diketahui kofaktor sebagai berikut :

$$A 
\begin{bmatrix}
4 & -1 \\
5 & -7
\end{bmatrix}$$

kemudian dilakukan transpose, menjadi :

$$A = 
\begin{bmatrix}
4 & 5 \\
-1 & -7
\end{bmatrix}$$

hitung invers :

= $A^{-1} = \frac{1}{-23} \begin{bmatrix} 4 & 5 \\ -1 & -7 \end{bmatrix}$

= $A^{-1} = \begin{bmatrix} -\frac{4}{23} & -\frac{5}{23} \\ \frac{1}{23} & \frac{7}{23} \end{bmatrix}$

maka inversnya adalah :

$\begin{bmatrix} -\frac{4}{23} & -\frac{5}{23} \\ \frac{1}{23} & \frac{7}{23} \end{bmatrix}$

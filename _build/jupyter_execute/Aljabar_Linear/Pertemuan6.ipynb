### Ekspansi Baris

jika matriksnya :

$$
\begin{bmatrix}
a&b\\
c&d\\
\end{bmatrix}
$$

$$\sum_{K=1}^{n} 1^{2+k}  \quad a_{2k} \quad M_{2k}$$

maka => $$(-1)^{2+1} \: a_{21} \: M_{21}$$ 
=> $$-cb + da$$
= $$ad - bc$$

Soal : Cari determinan A =  Ekspensi baris 2 dan 3

$$
\begin{bmatrix}
4&2&1\\
3&1&2\\
1&1&1
\end{bmatrix}
$$

Ekspansi baris 2 :

$$\sum_{K=1}^{3} (-1)^{2+k}  \quad a_{2k} \quad M_{2k}$$

maka :

$$(-1)^{2+1} \: a_{21} \: M_{21} + (-1)^{2+2} \: a_{22} \: M_{22} + (-1)^{2+3} \: a_{23} \: M_{23}$$

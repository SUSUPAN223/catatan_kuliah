soal 2

$$A=
\begin{bmatrix}
0 & 2 & -3 \\
1 & -2 & -1 \\
0 & 0 & 1
\end{bmatrix}
$$

jawaban :

Menggunakan ekspansi baris ke 3

$$
\sum_{j=1}^{n} (-1)^{3+j}a_{3j}det (A_{3j})
$$

kolom 1 :

$$+0
\begin{bmatrix}
2 & -3 \\
-2 & -1
\end{bmatrix}
= 2 \cdot -1 - (-3) \cdot -2 = -2 - 6 = -8 -> 0 \cdot -8 = 0
$$

kolom 2 :

$$-0
\begin{bmatrix}
0 & -3 \\
1 & -1
\end{bmatrix}
= 0 \cdot -1 - (-3) \cdot 1 = 0 - (-3) = 3 -> 0 \cdot 3 = 0
$$

kolom 3 :

$$+1
\begin{bmatrix}
0 & 2 \\
1 & -2
\end{bmatrix}
= 0 \cdot -2 - 2 \cdot 1 = 0 - 2 = -2
$$

hasil akhir :

$0-0+(-2) = -2$

Determinannya adalah -2

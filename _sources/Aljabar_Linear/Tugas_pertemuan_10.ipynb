```python
import matplotlib.pyplot as plt # Mengambil alat bantu untuk menggambar grafik di layar
import numpy as np # Mengambil alat bantu untuk berhitung angka dan mengatur bentuk kelompok data
import math # Mengambil alat hitung matematika tingkat lanjut untuk membuat efek bolak-balik (yoyo)
from matplotlib.animation import FuncAnimation # Mengambil alat untuk membuat kumpulan gambar menjadi video bergerak
from IPython.display import HTML # Mengambil alat untuk memutar video tersebut langsung di dalam Google Colab

A = [2, 3] # Menaruh ujung kiri atas kotak di posisi kanan 2 dan tinggi 3
B = [2, 4] # Menaruh ujung kanan atas kotak di posisi kanan 2 dan tinggi 4
C = [3, 4] # Menaruh ujung kanan bawah kotak di posisi kanan 3 dan tinggi 4
D = [3, 3] # Menaruh ujung kiri bawah kotak (kaki) di posisi kanan 3 dan tinggi 3
titik_awal = np.array([A, B, C, D, A], dtype=float) # Menyambungkan semua ujung tadi menjadi satu gambar kotak utuh

matriks_cermin = np.array([ # Membuka tempat untuk menaruh aturan cermin kita
    [1,  0], # Aturan baris pertama: biarkan posisi kiri dan kanan tetap di tempatnya
    [0, -1]  # Aturan baris kedua: paksa posisi atas menjadi bawah, dan bawah menjadi atas
]) # Menutup tempat aturan cermin

fig, ax = plt.subplots(figsize=(8, 8)) # Membentangkan kertas gambar kosong dengan ukuran persegi delapan lawan delapan

ax.spines['left'].set_position('zero') # Menggeser garis pengukur yang di kiri agar pindah ke tengah persis
ax.spines['bottom'].set_position('zero') # Menggeser garis pengukur yang di bawah agar naik ke tengah persis sebagai cermin
ax.spines['right'].set_color('none') # Menghapus garis pembatas kanan karena membuat pemandangan penuh
ax.spines['top'].set_color('none') # Menghapus garis pembatas atas agar gambarnya terlihat lega

ax.xaxis.set_ticks_position('bottom') # Memaksa tulisan angka jarak mendatar agar selalu menempel di garis tengah
ax.yaxis.set_ticks_position('left') # Memaksa tulisan angka tinggi agar selalu menempel di garis yang berdiri

plt.xlim(-8, 8) # Mengatur luas pandangan ke samping dari angka minus 8 sampai angka 8
plt.ylim(-10, 10) # Mengatur luas pandangan ke atas dan bawah dari angka minus 10 sampai 10 agar ruang geraknya luas

plt.grid(True, linestyle=':', alpha=0.7) # Menyalakan pola titik-titik samar di belakang agar terlihat seperti buku matematika
plt.title("Animasi Cermin Tembus Bolak-Balik") # Menuliskan judul besar di bagian paling atas kertas

garis_asli, = ax.plot([], [], 'b-o', label='Objek Asli (Biru)') # Menyiapkan spidol biru untuk menggambar kotak asli
garis_bayangan, = ax.plot([], [], 'r--o', label='Cerminan (Merah)') # Menyiapkan spidol merah putus-putus untuk bayangan
plt.legend(loc='upper left') # Memasang kotak petunjuk warna di pojokan kiri atas agar penonton paham

# --- BAGIAN BARU: MENGATUR EFEK BOLAK-BALIK ---
jarak_maksimal = 8 # Menentukan seberapa jauh kotak biru boleh turun tembus ke bawah sebelum disuruh putar balik
kecepatan = 0.1 # Menentukan seberapa cepat kotak tersebut bergerak naik turun
total_frame = int((2 * math.pi) / kecepatan) # Menghitung jumlah gambar yang pas untuk satu putaran gelombang penuh agar videonya nyambung terus (loop)
# ----------------------------------------------

def update(frame): # Membuat tugas rutin untuk menggambar ulang posisi kotak di setiap lembar gambar
    titik_bergerak = titik_awal.copy() # Memfotokopi kotak asli agar yang asli tidak rusak saat kita geser-geser
    
    # Menggunakan rumus gelombang (cosinus) agar gerakannya memantul lembut ujung ke ujung
    offset_turun = jarak_maksimal * (1 - math.cos(frame * kecepatan)) / 2 # Menghitung seberapa jauh kotak harus bergeser di detik ini
    
    titik_bergerak[:, 1] = titik_bergerak[:, 1] - offset_turun # Menurunkan posisi kotak biru sejauh hasil hitungan gelombang tadi
    
    titik_bayangan = titik_bergerak @ matriks_cermin # Menyuruh komputer menghitung posisi pantulan kotak merah menggunakan aturan cermin
    
    garis_asli.set_data(titik_bergerak[:, 0], titik_bergerak[:, 1]) # Menggambar ulang kotak biru di posisi barunya
    garis_bayangan.set_data(titik_bayangan[:, 0], titik_bayangan[:, 1]) # Menggambar ulang kotak merah di posisi barunya
    
    return garis_asli, garis_bayangan # Menyetorkan kedua gambar kotak tersebut ke layar untuk ditonton

animasi = FuncAnimation(fig, update, frames=total_frame, interval=50, blit=True) # Merangkai lembar gambar menjadi video yang berulang terus menerus

plt.close() # Menyembunyikan kertas coret-coretan diam agar tidak mengganggu video yang mau diputar

HTML(animasi.to_jshtml()) # Mengubah video tersebut menjadi tayangan web agar bisa langsung ditonton berulang-ulang di Colab
```

**Jawab :**

Step 1: Ubah titik ke dalam bentuk matriks

Rumus :

$$T(v_1, v_2) = (v_1 - v_2, v_1 + 2v_2)$$

titik a:

$$T(v_1, v_2) = (v_1 - v_2, v_1 + 2v_2)$$

maka $T(2, 3) = (2 - 3, 2 + 2(3))$ hasilnya : $T(2,3)=(-1,8)$

titik b:

$T(v_1, v_2) = (v_1 - v_2, v_1 + 2v_2)$

maka $T(2, 1) = (2 - 1, 2 + 2(1))$ hasilnya $T(2,1) = (1,4)$

titik c: 

$$T(v_1, v_2) = (v_1 - v_2, v_1 + 2v_2)$$

Maka $T(4, 1) = (4 - 1, 4 + 2(1))$ hasilnya $T(4,1) = (3, 6)$

titik d:

$$T(v_1, v_2) = (v_1 - v_2, v_1 + 2v_2)$$

Maka $T(2, 4) = (2 - 4, 2 + 2(4))$ hasilnya $T(2,4) = ( -2,10)$

titik e:

$$T(v_1, v_2) = (v_1 - v_2, v_1 + 2v_2)$$

Maka $T(2, 0) = (2 - 0, 2 + 2(0))$ hasilnya $T(2,0) = (2,2)$

titik f:

$$T(v_1, v_2) = (v_1 - v_2, v_1 + 2v_2)$$

Maka $T(4, 0) = (4 - 0, 4 + 2(0))$ hasilnya $T(4,0) = (4,4)$

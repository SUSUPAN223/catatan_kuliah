## Linear Interpolasi

Cara mencari sebuah linear interpolasi atau cara menghadapi missing values

Rumus Mencari Missing Values :

$$S_i(x) = y_i + p_i(x - x_i),$$


Lalu bagaimana mencari PI?

Rumus PI :$$p_i = \frac{\Delta y_i}{\Delta x_i}$$ 

Dimana Rumus Mencari delta yi adalah:

$$\Delta y_i \doteq y_{i+1} - y_i$$ 

Rumus Mencari Delta Xi :

$$\Delta x_i \doteq x_{i+1} - x_i .$$



## Interpolasi Polinomial

Rumus :

$$P(x) = \sum_{i=1}^{n} y_i \prod_{\substack{k=1 \\ k \neq i}}^{n} \frac{x - x_k}{x_i - x_k}.$$

```python
import ee
import folium 
import geemap 
import os

try:
    ee.Initialize()
except Exception as e:
    ee.Authenticate()
    ee.Initialize()

print("Earth Engine terhubung! Mulai memproses data...")

data_geojson = {
  "type": "FeatureCollection",
  "features": [
    {
      "type": "Feature",
      "properties": {"Class": "Sawah\n"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.7194401, -6.9000154], [107.7194455, -6.9004694], [107.7197645, -6.9008337], [107.7203589, -6.9001934], [107.7200837, -6.8994368], [107.7194401, -6.9000154]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.7194648, -6.902441], [107.7190107, -6.9029816], [107.7200259, -6.9038791], [107.7205297, -6.9029829], [107.7199001, -6.9020361], [107.7194648, -6.902441]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.7170611, -6.9052117], [107.7173069, -6.9048136], [107.7177467, -6.9047437], [107.7178042, -6.9051845], [107.7175797, -6.905598], [107.7173852, -6.9053261], [107.7170611, -6.9052117]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.716423, -6.9058063], [107.7166557, -6.9061675], [107.7168655, -6.9061381], [107.7166387, -6.9056764], [107.716423, -6.9058063]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.7147923, -6.905021], [107.7142921, -6.905795], [107.7139453, -6.9055903], [107.7147923, -6.905021]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.7195197, -6.9069888], [107.719482, -6.9072148], [107.7195946, -6.9073757], [107.7196683, -6.9073687], [107.7197714, -6.9071259], [107.7195197, -6.9069888]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.7197946, -6.9045892], [107.7199391, -6.9046773], [107.7198212, -6.9049566], [107.719648, -6.9048455], [107.7197946, -6.9045892]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.7208976, -6.9049312], [107.7208542, -6.9050113], [107.7211155, -6.9053562], [107.7211936, -6.9052773], [107.7208976, -6.9049312]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.7219176, -6.9046283], [107.7222266, -6.9048446], [107.722144, -6.9049485], [107.7218668, -6.9047633], [107.7219176, -6.9046283]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.7214399, -6.9005861], [107.7212181, -6.9008944], [107.7214727, -6.90113], [107.7217247, -6.9008258], [107.7214399, -6.9005861]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.7228387, -6.9017692], [107.722714, -6.9019961], [107.7233131, -6.9020731], [107.7228387, -6.9017692]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.7224688, -6.9022634], [107.722354, -6.9024312], [107.7225938, -6.9026627], [107.7227613, -6.9024534], [107.7224688, -6.9022634]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.7238108, -6.9022648], [107.7238556, -6.902328], [107.7236513, -6.9024753], [107.7235886, -6.9023995], [107.7238108, -6.9022648]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.7179948, -6.9093479], [107.7182384, -6.9095914], [107.7180135, -6.909881], [107.7177385, -6.9096776], [107.7179948, -6.9093479]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.7189582, -6.9094035], [107.7190853, -6.9095133], [107.7194014, -6.9092713], [107.7192634, -6.9091996], [107.7190615, -6.9092923], [107.7189582, -6.9094035]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.7254375, -6.9043562], [107.7253442, -6.9046177], [107.7255039, -6.9047472], [107.7256775, -6.9045106], [107.7254375, -6.9043562]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.7195082, -6.8983727], [107.7195596, -6.898863], [107.7197231, -6.8988951], [107.719757, -6.8987371], [107.7198769, -6.89865], [107.7196478, -6.8980152], [107.7195082, -6.8983727]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.7181901, -6.8981707], [107.7180675, -6.8985679], [107.7181503, -6.8986081], [107.7181755, -6.8984048], [107.7182697, -6.8982016], [107.7181901, -6.8981707]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.7186812, -6.8975925], [107.7185381, -6.8979229], [107.7186616, -6.897981], [107.7187832, -6.8976421], [107.7186812, -6.8975925]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.7201485, -6.8979211], [107.7202223, -6.8980911], [107.7202504, -6.8980276], [107.7201931, -6.8979364], [107.7201485, -6.8979211]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.720581, -6.8976989], [107.7207353, -6.8977988], [107.7207534, -6.897728], [107.7207482, -6.8976803], [107.7207707, -6.8976734], [107.7208061, -6.8977013], [107.720891, -6.8976261], [107.7207409, -6.8975615], [107.720581, -6.8976989]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.7130284, -6.9044109], [107.7130725, -6.9044966], [107.7128996, -6.9046352], [107.71286, -6.9046077], [107.7128295, -6.9046258], [107.7127865, -6.9045951], [107.7130284, -6.9044109]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.7126527, -6.903666], [107.7126103, -6.9036837], [107.7125608, -6.9038082], [107.7124537, -6.9038446], [107.7126299, -6.9039135], [107.7128615, -6.9039175], [107.7126527, -6.903666]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.7131581, -6.9024947], [107.7131786, -6.9025482], [107.7132962, -6.9025092], [107.7132938, -6.9024404], [107.7131581, -6.9024947]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.7132963, -6.9021713], [107.7132636, -6.9021915], [107.7132291, -6.9022708], [107.7134168, -6.902329], [107.7135675, -6.9023271], [107.7135399, -6.9022265], [107.7134141, -6.90223], [107.7132963, -6.9021713]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.7143012, -6.9033107], [107.7142334, -6.903492], [107.7143985, -6.9036018], [107.7144156, -6.9035772], [107.7145354, -6.9036305], [107.7145606, -6.9035356], [107.7145361, -6.9034775], [107.7146029, -6.9034094], [107.7144397, -6.903279], [107.7143012, -6.9033107]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.717451, -6.9028054], [107.7173995, -6.9028697], [107.7173661, -6.9029375], [107.717423, -6.9030813], [107.7174205, -6.9030425], [107.7175478, -6.9030472], [107.7175377, -6.9028851], [107.717451, -6.9028054]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.717808, -6.9029231], [107.7178221, -6.9029431], [107.7179413, -6.9028254], [107.7180265, -6.9026715], [107.7180028, -6.902652], [107.7179165, -6.9028028], [107.717808, -6.9029231]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.7182913, -6.9028538], [107.7182423, -6.9028776], [107.7182008, -6.9030158], [107.7183956, -6.9030629], [107.7185299, -6.9031466], [107.7185872, -6.9029764], [107.7184553, -6.9028556], [107.7182913, -6.9028538]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.7227636, -6.9007386], [107.7227091, -6.9007876], [107.7230587, -6.9010903], [107.7231393, -6.9010811], [107.7231313, -6.9010048], [107.7227636, -6.9007386]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.7237324, -6.9008419], [107.7236516, -6.9009726], [107.7236965, -6.9010035], [107.7237564, -6.9009928], [107.7238245, -6.9010729], [107.7238962, -6.9009865], [107.7237324, -6.9008419]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.7215569, -6.9081102], [107.7218891, -6.9082063], [107.7223241, -6.9079514], [107.7220947, -6.9079315], [107.7215569, -6.9081102]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.7228114, -6.9069838], [107.7227327, -6.907058], [107.7228704, -6.9071172], [107.7229242, -6.9070624], [107.7228114, -6.9069838]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.7228973, -6.9063396], [107.722927, -6.9063947], [107.7229155, -6.906465], [107.7228557, -6.9064925], [107.722826, -6.906441], [107.7228769, -6.9063809], [107.7228973, -6.9063396]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.7227806, -6.9057109], [107.7227647, -6.9058365], [107.7228052, -6.9058476], [107.7229524, -6.9057553], [107.7230055, -6.9057059], [107.7228296, -6.9056814], [107.7227806, -6.9057109]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.7280691, -6.9049856], [107.7281687, -6.9050662], [107.7278602, -6.9054048], [107.7276668, -6.9052556], [107.7278156, -6.9051044], [107.7278859, -6.9051501], [107.7280691, -6.9049856]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.727839, -6.9039579], [107.7279032, -6.9040719], [107.727854, -6.9041967], [107.7277271, -6.9040063], [107.727839, -6.9039579]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.7284932, -6.9031251], [107.7282651, -6.9032713], [107.7283929, -6.9034781], [107.7286294, -6.9033359], [107.7284932, -6.9031251]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.7229178, -6.8999697], [107.7228797, -6.9000542], [107.7231399, -6.9002399], [107.7231794, -6.9001526], [107.7229178, -6.8999697]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.7249564, -6.8993145], [107.7248119, -6.8994439], [107.7248866, -6.8994818], [107.72502, -6.8993815], [107.7249564, -6.8993145]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.7255629, -6.8990027], [107.7256579, -6.8990644], [107.7256365, -6.8990993], [107.7257073, -6.8991436], [107.725576, -6.8993205], [107.7254242, -6.8991887], [107.7254883, -6.8990557], [107.7255629, -6.8990027]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.7269875, -6.8995549], [107.7269515, -6.8997292], [107.7271777, -6.8997852], [107.7272409, -6.8997319], [107.7273019, -6.899727], [107.7274105, -6.8995946], [107.7272397, -6.8994965], [107.7269875, -6.8995549]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.7218865, -6.8962917], [107.7219446, -6.8963636], [107.7219459, -6.8964581], [107.7222441, -6.8967939], [107.7220881, -6.8970364], [107.7216653, -6.8967301], [107.7214372, -6.8963951], [107.7218865, -6.8962917]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.7198445, -6.8959083], [107.7199686, -6.8960018], [107.7199343, -6.8960567], [107.7198619, -6.8960274], [107.719796, -6.8960842], [107.719742, -6.8960471], [107.7198445, -6.8959083]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.720265, -6.895409], [107.7203006, -6.895457], [107.7202291, -6.8955423], [107.7202219, -6.8955062], [107.7202415, -6.89548], [107.7202196, -6.8954618], [107.720265, -6.895409]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.7212979, -6.8953724], [107.7212831, -6.895428], [107.7211658, -6.8954881], [107.7211365, -6.8954568], [107.7212173, -6.8954253], [107.7212979, -6.8953724]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.7204044, -6.8948922], [107.720604, -6.8948948], [107.7206828, -6.8949293], [107.7207784, -6.8950126], [107.7207505, -6.8949478], [107.7206167, -6.8948592], [107.7204243, -6.894847], [107.7204044, -6.8948922]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.7201616, -6.8971706], [107.7202185, -6.8972742], [107.7202106, -6.89732], [107.7202433, -6.8973209], [107.7203009, -6.8972415], [107.7202422, -6.8972312], [107.7202194, -6.8971288], [107.7201616, -6.8971706]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.7231815, -6.8958554], [107.7232444, -6.8958791], [107.7232582, -6.8959205], [107.7233091, -6.895924], [107.7233518, -6.8960027], [107.7232738, -6.896076], [107.7232344, -6.8960346], [107.7232378, -6.8959566], [107.7231718, -6.8959946], [107.7231239, -6.8959463], [107.7231815, -6.8958554]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.7273998, -6.8988737], [107.727364, -6.89894], [107.7274431, -6.8990537], [107.7274889, -6.8990032], [107.7274496, -6.8989213], [107.7273998, -6.8988737]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Bukan Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.716538, -6.9064212], [107.7162128, -6.906765], [107.7163564, -6.9068824], [107.7169645, -6.9065721], [107.7173277, -6.9064883], [107.7171546, -6.9063332], [107.716804, -6.906417], [107.716538, -6.9064212]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Bukan Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.717225, -6.9074461], [107.717696, -6.9078515], [107.7174588, -6.9081523], [107.7170209, -6.9076979], [107.717225, -6.9074461]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Bukan Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.7148619, -6.9082563], [107.7144955, -6.9089838], [107.7150634, -6.9091839], [107.715155, -6.9088323], [107.7154054, -6.9085594], [107.7148619, -6.9082563]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Bukan Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.7126856, -6.90889], [107.7133374, -6.9101295], [107.7137689, -6.9094642], [107.7133374, -6.9084253], [107.7126856, -6.90889]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Bukan Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.7087962, -6.9092664], [107.7097618, -6.9096468], [107.7095932, -6.9100728], [107.7088115, -6.9096164], [107.7087962, -6.9092664]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Bukan Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.7156553, -6.9107504], [107.7158467, -6.9113773], [107.7163251, -6.9110353], [107.7156553, -6.9107504]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Bukan Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.7155979, -6.9139422], [107.7161912, -6.9143792], [107.7167462, -6.9139802], [107.7159998, -6.9136192], [107.7155979, -6.9139422]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Bukan Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.7190045, -6.9145502], [107.7187366, -6.9139042], [107.7196935, -6.9137712], [107.7197892, -6.9145122], [107.7190045, -6.9145502]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Bukan Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.7166577, -6.9011315], [107.7165246, -6.9014901], [107.7168858, -6.9017543], [107.7171139, -6.9012825], [107.7166577, -6.9011315]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Bukan Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.7102171, -6.90645], [107.7100822, -6.9073459], [107.711078, -6.90749], [107.7112232, -6.9062647], [107.7102171, -6.90645]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Bukan Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.7119907, -6.90749], [107.7117937, -6.9079122], [107.7120841, -6.9083138], [107.7125612, -6.9075003], [107.7119907, -6.90749]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Bukan Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.7089809, -6.9015399], [107.7097358, -6.9018829], [107.7096846, -6.9022512], [107.708789, -6.9018829], [107.7089809, -6.9015399]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Bukan Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.708036, -6.9040022], [107.7082412, -6.9045506], [107.7079729, -6.9055064], [107.7076572, -6.9051304], [107.7074678, -6.904488], [107.708036, -6.9040022]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Bukan Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.70596, -6.9060184], [107.7063682, -6.9068635], [107.7055402, -6.9071645], [107.7051671, -6.9066783], [107.70596, -6.9060184]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Bukan Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.7117762, -6.8976334], [107.7116596, -6.8981348], [107.7122683, -6.8984948], [107.7127605, -6.8981348], [107.7117762, -6.8976334]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Bukan Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.7104447, -6.9000341], [107.7105137, -6.9005273], [107.711045, -6.9004999], [107.7107138, -6.8997395], [107.7104447, -6.9000341]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Bukan Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.7071391, -6.8992781], [107.7069597, -6.8998604], [107.707215, -6.8999563], [107.7076566, -6.8992096], [107.7071391, -6.8992781]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Bukan Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.705071, -6.8998195], [107.7050915, -6.9005165], [107.7055618, -6.900706], [107.7057731, -6.8999616], [107.705071, -6.8998195]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Bukan Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.7056871, -6.9023524], [107.7058181, -6.9028673], [107.7065735, -6.9027073], [107.7064174, -6.9020424], [107.7056871, -6.9023524]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Bukan Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.7046723, -6.9033348], [107.705186, -6.9034298], [107.7054026, -6.9030399], [107.7048083, -6.9029449], [107.7046723, -6.9033348]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Bukan Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.7034711, -6.9047684], [107.7032003, -6.9055994], [107.704062, -6.9057786], [107.7042097, -6.9048173], [107.7034711, -6.9047684]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Bukan Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.7014808, -6.9034326], [107.7013995, -6.9038727], [107.7020349, -6.9040414], [107.7024635, -6.9035793], [107.7014808, -6.9034326]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Bukan Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.700179, -6.9055078], [107.6999648, -6.9061899], [107.7004007, -6.9063146], [107.700615, -6.9060579], [107.7005559, -6.9056178], [107.700179, -6.9055078]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Bukan Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.6993921, -6.9074565], [107.7002064, -6.9078634], [107.7003359, -6.9077724], [107.7003682, -6.9075315], [107.6995161, -6.9071835], [107.6993921, -6.9074565]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Bukan Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.7141204, -6.8953027], [107.7143512, -6.8966056], [107.7147045, -6.8966844], [107.7152958, -6.895904], [107.7141204, -6.8953027]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Bukan Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.7170949, -6.8947583], [107.7173117, -6.8957592], [107.7180814, -6.895813], [107.7179188, -6.8954578], [107.7181247, -6.8949628], [107.7179404, -6.8942848], [107.7170949, -6.8947583]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Bukan Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.7072407, -6.8955084], [107.706981, -6.8958887], [107.7070914, -6.896501], [107.7082795, -6.8965526], [107.7083769, -6.8960047], [107.7072407, -6.8955084]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Bukan Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.7036721, -6.897721], [107.7036504, -6.8983022], [107.7044309, -6.8980654], [107.7036721, -6.897721]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Bukan Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.7247031, -6.8939989], [107.724245, -6.8943611], [107.7250171, -6.8946391], [107.7247031, -6.8939989]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Bukan Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.7254614, -6.8927756], [107.7258194, -6.8932808], [107.726121, -6.8924014], [107.7257723, -6.8923172], [107.7254614, -6.8927756]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Bukan Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.7205295, -6.891826], [107.720289, -6.8929941], [107.7209932, -6.8931475], [107.7215343, -6.8919453], [107.7205295, -6.891826]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Bukan Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.7194942, -6.890411], [107.7204406, -6.8894003], [107.7194942, -6.8892722], [107.7194942, -6.890411]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Bukan Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.7059654, -6.9080737], [107.7063537, -6.9099727], [107.7047429, -6.90963], [107.7059654, -6.9080737]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Bukan Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.7024824, -6.910115], [107.7029301, -6.9110039], [107.7024045, -6.9121248], [107.7013923, -6.9103469], [107.7024824, -6.910115]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Bukan Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.7077658, -6.9143244], [107.7083442, -6.9156467], [107.7070472, -6.9160817], [107.7068368, -6.9152117], [107.7077658, -6.9143244]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Bukan Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.6902511, -6.9092771], [107.6890566, -6.9109318], [107.691958, -6.9113102], [107.6924025, -6.9101244], [107.6902511, -6.9092771]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Bukan Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.6970417, -6.9134061], [107.6972361, -6.9155296], [107.6989029, -6.9151711], [107.6975139, -6.9129649], [107.6970417, -6.9134061]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Bukan Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.6834853, -6.9063187], [107.6855132, -6.9073115], [107.6829019, -6.9111999], [107.6820686, -6.9106484], [107.681513, -6.9066772], [107.6834853, -6.9063187]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Bukan Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.6916803, -6.9007755], [107.6893468, -6.9037539], [107.6922636, -6.9043606], [107.6916803, -6.9007755]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Bukan Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.6958185, -6.8979622], [107.696175, -6.8992192], [107.6970478, -6.8984626], [107.6958185, -6.8979622]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Bukan Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.6966333, -6.8934051], [107.6956253, -6.8945889], [107.6971988, -6.8952723], [107.6977766, -6.8938444], [107.6966333, -6.8934051]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Bukan Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.6999752, -6.8909251], [107.7009095, -6.8913157], [107.7006513, -6.8922676], [107.6997293, -6.8922676], [107.6999752, -6.8909251]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Bukan Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.7002017, -6.8991], [107.7014342, -6.8994345], [107.6997673, -6.900702], [107.7002017, -6.8991]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Bukan Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.7006108, -6.8962127], [107.7014265, -6.8970226], [107.7008679, -6.8976035], [107.6999458, -6.897049], [107.7006108, -6.8962127]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Bukan Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.7040811, -6.8931461], [107.7044269, -6.8940967], [107.704888, -6.8938679], [107.7048525, -6.8931373], [107.7040811, -6.8931461]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Bukan Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.6954864, -6.9071273], [107.6951916, -6.9079541], [107.6961572, -6.9082395], [107.696194, -6.9071565], [107.6954864, -6.9071273]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Bukan Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.6976771, -6.9024878], [107.6981963, -6.904406], [107.6960407, -6.9038833], [107.6976204, -6.9025764], [107.6976771, -6.9024878]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Bukan Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.6912183, -6.8976718], [107.6923443, -6.8980836], [107.6907053, -6.8989828], [107.6910609, -6.8980414], [107.6911794, -6.8977473], [107.6912183, -6.8976718]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Bukan Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.6938903, -6.89527], [107.6953916, -6.8961762], [107.694719, -6.896999], [107.6933498, -6.8962239], [107.6938903, -6.89527]]]}
    },
    {
      "type": "Feature",
      "properties": {"Class": "Bukan Sawah"},
      "geometry": {"type": "Polygon", "coordinates": [[[107.7032616, -6.8950115], [107.7038135, -6.8954714], [107.7033995, -6.8963031], [107.7025618, -6.8958726], [107.7032616, -6.8950115]]]}
    }
  ]
}

fitur_list = []

kelas_unik = list(set([f['properties']['Class'].strip() for f in data_geojson['features']]))
kelas_unik.sort() 
mapping_kelas = {kelas: i for i, kelas in enumerate(kelas_unik)}
print(f"Ditemukan {len(kelas_unik)} kelas otomatis dari data: {mapping_kelas}")

for feature in data_geojson['features']:
    coords = feature['geometry']['coordinates']
    geom = ee.Geometry.Polygon(coords)
    
    kelas_str = feature['properties']['Class'].strip()
    kelas_val = mapping_kelas[kelas_str]
    
    fitur_list.append(ee.Feature(geom, {'Classe': kelas_val}))

data_training = ee.FeatureCollection(fitur_list)

Batas_Area = data_training.geometry().bounds()

Sentinel2 = ee.ImageCollection("COPERNICUS/S2_SR_HARMONIZED")
image_s2 = ee.Image(Sentinel2.filterDate("2023-01-01", "2023-12-31").filterBounds(Batas_Area).sort("CLOUDY_PIXEL_PERCENTAGE").first()) 
S2_fill = image_s2.focal_mean(1, 'square', 'pixels', 9)
S2_final = S2_fill.blend(image_s2).clip(Batas_Area)

ndvi = S2_final.normalizedDifference(['B8', 'B4']).rename('ndvi')
ndwi = S2_final.normalizedDifference(['B3', 'B8']).rename('ndwi')
S2_lengkap = S2_final.addBands([ndvi, ndwi])

bandes = ['B2', 'B3', 'B4', 'B8', 'ndvi', 'ndwi']
entrainement = S2_lengkap.select(bandes).sampleRegions(collection=data_training, properties=['Classe'], scale=10)
classicateur = ee.Classifier.smileRandomForest(100).train(features=entrainement, classProperty='Classe', inputProperties=bandes)
hasil_klasifikasi = S2_lengkap.select(bandes).classify(classicateur)

print("Klasifikasi selesai! Menyiapkan peta...")

def add_ee_layer(self, ee_image_object, vis_params, name):
    map_id_dict = ee.Image(ee_image_object).getMapId(vis_params)
    folium.raster_layers.TileLayer(
        tiles=map_id_dict['tile_fetcher'].url_format,
        attr='Map Data &copy; Google Earth Engine',
        name=name,
        overlay=True,
        control=True
    ).add_to(self)

folium.Map.add_ee_layer = add_ee_layer

titik_tengah = Batas_Area.centroid(1).coordinates().getInfo()
peta_saya = folium.Map(location=[titik_tengah[1], titik_tengah[0]], zoom_start=15)

peta_saya.add_ee_layer(image_s2, {'bands': ['B4', 'B3', 'B2'], 'min': 0, 'max': 3000}, 'Citra Sentinel-2')

daftar_warna = ['red', 'green', 'blue', 'orange', 'purple']
warna_dipakai = daftar_warna[:len(kelas_unik)]

peta_saya.add_ee_layer(
    hasil_klasifikasi, 
    {'min': 0, 'max': len(kelas_unik) - 1, 'palette': warna_dipakai}, 
    'Hasil Klasifikasi Lahan'
)
peta_saya.add_child(folium.LayerControl())

legend_html = f"""
<div style="position: fixed; bottom: 50px; left: 50px; width: 160px; height: auto; 
            border:2px solid grey; z-index:9999; font-size:14px; background-color:white; padding: 10px;">
<b>Keterangan Lahan:</b><br>
"""
for i, kelas in enumerate(kelas_unik):
    legend_html += f'<i style="background:{warna_dipakai[i]}; width: 12px; height: 12px; float: left; margin-right: 8px; margin-top: 3px;"></i> {kelas}<br>'
legend_html += "</div>"

peta_saya.get_root().html.add_child(folium.Element(legend_html))
display(peta_saya)

nama_file = 'Klasifikasi_Lahan_Cilengkrang_Bandung.tif'
print(f"\nSedang mendownload {nama_file}...")
geemap.ee_export_image(hasil_klasifikasi, filename=nama_file, scale=10, region=Batas_Area, file_per_band=False)
print("Selesai!")
```

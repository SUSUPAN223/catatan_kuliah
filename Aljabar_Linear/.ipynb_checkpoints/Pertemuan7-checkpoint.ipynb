Sebuah determinan bisa di ketahui dari baris 1 dan kolom 1 di kalikan


Soal :

mencari invers
syaratnya adalah determinan tidak boleh = 0
jadi kita mencari determinannya dulu

$$\begin{aligned}
x + y + z + m &= 10 \\
2x + -y + z + -m &= -1 \\
x + -2y + -z + m &= 6 \\
3x + -y + 2z + m &= 11
\end{aligned}$$

$$
\begin{bmatrix}
1&1&1&1&|&10\\
2&-1&1&-1&|&-1\\
1&-2&-1&1&|&6 \\
3&-1&2&1&|&11
\end{bmatrix}
$$

jawaban :
Yang pertama kali kita lakukan adalah mencari determinannya

rumus determinan adalah 

$$
\sum_{j=1}^{n} (-1)^{1+j}a_{1j}det (A_{1j})
$$

Kita akan melakukan ekspansi baris

maka, a_{1j} disini berfungsi untuk mengambil nilai pada matrix baris 1 kolom j, yang jika diimplementasikan pada soal kolom 1, dia akan mengambil angka 1

kemudian A_{1j} adalah sub matriks yang terbentuk setelah menghapus baris 1 kolom j, nilai dari sub matriks inilah yang disebut dengan minor(M)

implementasi rumus ke soal

$$
D = (-1)^{1+1}a_{11}\det(A_{11}) + (-1)^{1+2}a_{12}\det(A_{12}) + (-1)^{1+3}a_{13}\det(A_{13}) + (-1)^{1+4}a_{14}\det(A_{14})
$$

penjelasan :

mengapa di awal berbentuk seperti ini  $D=(-1)^{1+1}a_{11}det(A_{11})$?   karna kita memulai dari kolom pertama baris ke 1, yang artinya j=1 untuk putaran pertama, maka $(-1)^{1+j}$ akan menjadi $(-1)^{1+1}$

setelah terbentuk determinan ini, kita hitung $(-1)^{1+j}a_{1j}$ nya secara rinci, cara hitungnya :

pertama, hitung pangkat $^{1+j}$, karna putaran pertama j=1 (kolom 1), maka $^{1+j}$ akan menjadi $^{1+1}$ yang hasilnya adalah 2, berarti posisi saat ini adalah $(-1)^2$. Perlu diperhatikan, jika i+j hasilnya adalah angka genap $(-1)^2$ seperti ini, maka angka -1 di awal akan di ubah menjadi (+1)

kedua, hitung $a_{1j}$, karna j=1 (kolom 1) maka $a_{1j}$ menjadi $a_{11}$ yang berarti mengambil angka yang ada di baris 1 kolom 1, pada soal angkanya adalah 1

ketiga, setelah ketemu $(-1)^2$ adalah (+1) dan $a_{11}$ adalah 1, maka keduanya di kalikan jadi $(+1) \cdot 1$ maka hasilnya adalah +1.

keempat, setelah langkah 1 - 3 selesai, tinggal merubah $\det(A_{11})$, caranya? $\det(A_{11})$ ini adalah sebuah fungsi yang akan menghapus baris 1 kolom ke 1, sehingga matrix yang awalnya berukuran 4x4 karna dihapus baris 1 dan kolom 1 nya, berubah menjadi matrix 3x3, nah nilai dari deteriminan ini, itulah yang disebut dengan $MINOR$

hasil dari keempat langkah tersebut, terbentuklah persamaan seperti ini:

$$D = 
\underbrace{+1 \begin{vmatrix} -1 & 1 & -1 \\ -2 & -1 & 1 \\ -1 & 2 & 1 \end{vmatrix}}_{j=1} 
\underbrace{-1 \begin{vmatrix} 2 & 1 & -1 \\ 1 & -1 & 1 \\ 3 & 2 & 1 \end{vmatrix}}_{j=2} 
\underbrace{+1 \begin{vmatrix} 2 & -1 & -1 \\ 1 & -2 & 1 \\ 3 & -1 & 1 \end{vmatrix}}_{j=3} 
\underbrace{-1 \begin{vmatrix} 2 & -1 & 1 \\ 1 & -2 & -1 \\ 3 & -1 & 2 \end{vmatrix}}_{j=4}
$$

setelah menjadi persamaan, kita kembali menghitung menggunakan rumus determinan lagi, pada minor minor dari determinan 4x4 tadi :

$$
\sum_{k=1}^{n} (-1)^{1+k}a_{1k}det (A_{1k})
$$


hitung ulang :

pertama, hitung pangkat $^{1+k}$, karna putaran pertama k=1 (kolom 1), maka $^{1+k}$ akan menjadi $^{1+1}$ yang hasilnya adalah 2, berarti posisi saat ini adalah $(-1)^2$. Perlu diperhatikan, jika i+k hasilnya adalah angka genap $(-1)^2$ seperti ini, maka angka -1 di awal akan di ubah menjadi (+1)

kedua, hitung $a_{1k}$, karna k=1 (kolom 1) maka $a_{1k}$ menjadi $a_{11}$ yang berarti mengambil angka yang ada di baris 1 kolom 1, pada tahap ini angkanya adalah -1

ketiga, setelah ketemu $(-1)^2$ adalah (+1) dan $a_{11}$ adalah -1, maka keduanya di kalikan jadi $(+1) \cdot (-1)$ maka hasilnya adalah -1.

keempat, setelah langkah 1 - 3 selesai, tinggal merubah $\det(A_{11})$, caranya? $\det(A_{11})$ ini adalah sebuah fungsi yang akan menghapus baris 1 kolom ke 1, sehingga submatrix yang awalnya berukuran 3x3 karna dihapus baris 1 dan kolom 1 nya, berubah menjadi matrix 2x2.

hasil dari keempat langkah ini adalah:

$$D = +1 \left[ 
\underbrace{-1 \begin{vmatrix} -1 & 1 \\ 2 & 1 \end{vmatrix}}_{k=1}  
\underbrace{-1 \begin{vmatrix} -2 & 1 \\ -1 & 1 \end{vmatrix}}_{k=2}  
\underbrace{-1 \begin{vmatrix} -2 & -1 \\ -1 & 2 \end{vmatrix}}_{k=3} 
\right]
-1 \left[ 
\underbrace{+(2) \begin{vmatrix} -2 & 1 \\ -1 & 1 \end{vmatrix}}_{k=1} 
\underbrace{-(1) \begin{vmatrix} 1 & 1 \\ 3 & 1 \end{vmatrix}}_{k=2} 
\underbrace{+(-1) \begin{vmatrix} 1 & -2 \\ 3 & 2 \end{vmatrix}}_{k=3} 
\right]
+1 \left[ 
\underbrace{+(2) \begin{vmatrix} -2 & 1 \\ -1 & 1 \end{vmatrix}}_{k=1} 
\underbrace{-(-1) \begin{vmatrix} 1 & 1 \\ 3 & 1 \end{vmatrix}}_{k=2} 
\underbrace{+(-1) \begin{vmatrix} 1 & -2 \\ 3 & -1 \end{vmatrix}}_{k=3} 
\right]
-1 \left[ 
\underbrace{+(2) \begin{vmatrix} -2 & -1 \\ -1 & 2 \end{vmatrix}}_{k=1} 
\underbrace{-(-1) \begin{vmatrix} 1 & -1 \\ 3 & 2 \end{vmatrix}}_{k=2} 
\underbrace{+(1) \begin{vmatrix} 1 & -2 \\ 3 & -1 \end{vmatrix}}_{k=3} 
\right]$$

setelah menjadi determinan seperti ini, kita lanjut menggunakan rumus ini pada masing masing Minor nya:

$$\det \begin{bmatrix} a & b \\ c & d \end{bmatrix} = ad - bc$$

yang perlu dilakukan, hanyalah melakukan perkalian silang antara a dan d, kemudian mengalikan b dan c, lalu hasilnya di kurangi, posisi saat ini adalah sebagai berikut :



$$\begin{aligned}
D = \\
&\mathbf{+1} \begin{bmatrix} \begin{aligned}
&-1 \left[ \begin{bmatrix} -1 & 1 \\ 2 & 1 \end{bmatrix} = -1 \cdot 1 - 1 \cdot 2 = -3 \right] \\
&-1 \left[ \begin{bmatrix} -2 & 1 \\ -1 & 1 \end{bmatrix} = -2 \cdot 1 - 1 \cdot (-1) = -1 \right] \\
&-1 \left[ \begin{bmatrix} -2 & -1 \\ -1 & 2 \end{bmatrix} = -2 \cdot (-1) - (-1) \cdot (-1) = 1 \right] 
\end{aligned} \end{bmatrix} \\
\\
&\mathbf{-1} \begin{bmatrix} \begin{aligned}
&+2 \left[ \begin{bmatrix} -2 & 1 \\ -1 & 1 \end{bmatrix} = -2 \cdot 1 - 1 \cdot (-1) = -1 \right] \\
&-1 \left[ \begin{bmatrix} 1 & 1 \\ 3 & 1 \end{bmatrix} = 1 \cdot 1 - 1 \cdot 3 = -2 \right] \\
&-1 \left[ \begin{bmatrix} 1 & -2 \\ 3 & 2 \end{bmatrix} = 1 \cdot 2 - (-2) \cdot 3 = 8 \right]
\end{aligned} \end{bmatrix} \\
\\
&\mathbf{+1} \begin{bmatrix} \begin{aligned}
&+2 \left[ \begin{bmatrix} -2 & 1 \\ -1 & 1 \end{bmatrix} = -2 \cdot 1 - 1 \cdot (-1) = -1 \right] \\
&+1 \left[ \begin{bmatrix} 1 & 1 \\ 3 & 1 \end{bmatrix} = 1 \cdot 1 - 1 \cdot 3 = -2 \right] \\
&-1 \left[ \begin{bmatrix} 1 & -2 \\ 3 & -1 \end{bmatrix} = 1 \cdot (-1) - (-2) \cdot 3 = 5 \right]
\end{aligned} \end{bmatrix} \\
\\
&\mathbf{-1} \begin{bmatrix} \begin{aligned}
&+2 \left[ \begin{bmatrix} -2 & -1 \\ -1 & 2 \end{bmatrix} = -2 \cdot 2 - (-1) \cdot (-1) = -5 \right] \\
&+1 \left[ \begin{bmatrix} 1 & -1 \\ 3 & 2 \end{bmatrix} = 1 \cdot 2 - (-1) \cdot 3 = 5 \right] \\
&+1 \left[ \begin{bmatrix} 1 & -2 \\ 3 & -1 \end{bmatrix} = 1 \cdot (-1) - (-2) \cdot 3 = 5 \right]
\end{aligned} \end{bmatrix}
\end{aligned}$$


Setelah sudah di hitung per masing masing matrix 2x2 nya, pada soal ini berjumlah 12 matrix 2x2, maka dilanjutkan dengan mengalikan angka sebelum minor, dengan hasil matrix 2x2 

seperti ini:

$+1 [ (-1 \cdot -3) + (-1 \cdot -1) + (-1 \cdot 1) ] = 1[3 + 1 - 1] = \mathbf{3}$

$-1 [ (2 \cdot -1) + (-1 \cdot -2) + (-1 \cdot 8) ] = -1[-2 + 2 - 8] = \mathbf{8}$

$+1 [ (2 \cdot -1) + (1 \cdot -2) + (-1 \cdot 5) ] = 1[-2 - 2 - 5] = \mathbf{-9}$

$-1 [ (2 \cdot -5) + (1 \cdot 5) + (1 \cdot 5) ] = -1[-10 + 5 + 5] = \mathbf{0}$

Hasil Akhir Determinan Utama ($D$):
$$D = 3 + 8 - 9 + 0 = \mathbf{2}$$

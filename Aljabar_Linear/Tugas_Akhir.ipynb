## A. Soal Determinan Matriks 3×3

### Soal 1
$$
A=
\begin{bmatrix}
2 & 1 & 3 \\
0 & 4 & 5 \\
1 & 2 & 1
\end{bmatrix}
$$


**Jawaban :** 

Menggunakan ekspansi baris ke 1

$$\sum_{j=1}^{3} (-1)^{1+j}a_{1j}\det (A_{1j})$$

**Kolom 1:**
$$+2 \begin{bmatrix} 4 & 5 \\ 2 & 1 \end{bmatrix} = 2 \cdot (4 \cdot 1 - 5 \cdot 2) = 2 \cdot (4 - 10) = 2 \cdot (-6) = -12$$

**Kolom 2:**
$$-1 \begin{bmatrix} 0 & 5 \\ 1 & 1 \end{bmatrix} = -1 \cdot (0 \cdot 1 - 5 \cdot 1) = -1 \cdot (0 - 5) = -1 \cdot (-5) = 5$$

**Kolom 3:**
$$+3 \begin{bmatrix} 0 & 4 \\ 1 & 2 \end{bmatrix} = 3 \cdot (0 \cdot 2 - 4 \cdot 1) = 3 \cdot (0 - 4) = 3 \cdot (-4) = -12$$

**Hasil Akhir:**
$$-12 + 5 + (-12) = -19$$

Determinannya adalah **-19**.


### Soal 2

$$B=
\begin{bmatrix}
3 & 2 & 1 \\
1 & 0 & 4 \\
2 & 5 & 1
\end{bmatrix}
$$

**Jawaban :**

Menggunakan ekspansi baris ke 1

**Rumus Umum Ekspansi Kofaktor (Baris 1):**
$$\sum_{j=1}^{3} (-1)^{1+j}b_{1j}\det (B_{1j})$$

**Kolom 1:**
$$+3 \begin{bmatrix} 0 & 4 \\ 5 & 1 \end{bmatrix} = 3 \cdot (0 \cdot 1 - 4 \cdot 5) = 3 \cdot (0 - 20) = 3 \cdot (-20) = -60$$

**Kolom 2:**
$$-2 \begin{bmatrix} 1 & 4 \\ 2 & 1 \end{bmatrix} = -2 \cdot (1 \cdot 1 - 4 \cdot 2) = -2 \cdot (1 - 8) = -2 \cdot (-7) = 14$$

**Kolom 3:**
$$+1 \begin{bmatrix} 1 & 0 \\ 2 & 5 \end{bmatrix} = 1 \cdot (1 \cdot 5 - 0 \cdot 2) = 1 \cdot (5 - 0) = 1 \cdot 5 = 5$$

**Hasil Akhir:**
$$-60 + 14 + 5 = -41$$

Determinannya adalah **-41**.

## B. Soal Dekomposisi Matriks (LU Decomposition)

### Soal 4

$$A=\begin{bmatrix}2 & 4 & 2 \\ 1 & 5 & 2 \\ 1 & 2 & 4\end{bmatrix}$$

**Jawaban:**

**Kolom 1:**
* Pengali $m_{21}$:
  $$m_{21} = \frac{1}{2}$$
* Operasi Baris Elemen (OBE) $B_2$:
  $$B_2 \leftarrow B_2 - \frac{1}{2}B_1 = [1, 5, 2] - [1, 2, 1] = [0, 3, 1]$$
* Pengali $m_{31}$:
  $$m_{31} = \frac{1}{2}$$
* Operasi Baris Elemen (OBE) $B_3$:
  $$B_3 \leftarrow B_3 - \frac{1}{2}B_1 = [1, 2, 4] - [1, 2, 1] = [0, 0, 3]$$

**Kolom 2:**
* Pengali $m_{32}$:
  $$m_{32} = \frac{0}{3} = 0$$
* Operasi Baris Elemen (OBE) $B_3$:
  $$B_3 \leftarrow B_3 - 0 B_2 = [0, 0, 3] - [0, 0, 0] = [0, 0, 3]$$

**Hasil Akhir (Faktorisasi LU):**
$$L = \begin{bmatrix} 1 & 0 & 0 \\ \frac{1}{2} & 1 & 0 \\ \frac{1}{2} & 0 & 1 \end{bmatrix}, \quad U = \begin{bmatrix} 2 & 4 & 2 \\ 0 & 3 & 1 \\ 0 & 0 & 3 \end{bmatrix}$$

## C. Soal Invers Matriks 3×3

### Soal 7
$$A=\begin{bmatrix}1 & 2 & 1 \\ 0 & 1 & 1 \\ 2 & 3 & 4\end{bmatrix}$$

**Jawaban:**

#### 1. Menghitung Determinan (Ekspansi Baris 1)
$$\det(A) = 1 \begin{vmatrix} 1 & 1 \\ 3 & 4 \end{vmatrix} - 2 \begin{vmatrix} 0 & 1 \\ 2 & 4 \end{vmatrix} + 1 \begin{vmatrix} 0 & 1 \\ 2 & 3 \end{vmatrix}$$
$$\det(A) = 1(4 - 3) - 2(0 - 2) + 1(0 - 2) = 1(1) - 2(-2) + 1(-2) = 1 + 4 - 2 = 3$$

#### 2. Menghitung Kofaktor
* **Baris 1:**
  $$C_{11} = + \begin{vmatrix} 1 & 1 \\ 3 & 4 \end{vmatrix} = 4 - 3 = 1$$
  $$C_{12} = - \begin{vmatrix} 0 & 1 \\ 2 & 4 \end{vmatrix} = -(0 - 2) = 2$$
  $$C_{13} = + \begin{vmatrix} 0 & 1 \\ 2 & 3 \end{vmatrix} = 0 - 2 = -2$$

* **Baris 2:**
  $$C_{21} = - \begin{vmatrix} 2 & 1 \\ 3 & 4 \end{vmatrix} = -(8 - 3) = -5$$
  $$C_{22} = + \begin{vmatrix} 1 & 1 \\ 2 & 4 \end{vmatrix} = 4 - 2 = 2$$
  $$C_{23} = - \begin{vmatrix} 1 & 2 \\ 2 & 3 \end{vmatrix} = -(3 - 4) = 1$$

* **Baris 3:**
  $$C_{31} = + \begin{vmatrix} 2 & 1 \\ 1 & 1 \end{vmatrix} = 2 - 1 = 1$$
  $$C_{32} = - \begin{vmatrix} 1 & 1 \\ 0 & 1 \end{vmatrix} = -(1 - 0) = -1$$
  $$C_{33} = + \begin{vmatrix} 1 & 2 \\ 0 & 1 \end{vmatrix} = 1 - 0 = 1$$

#### 3. Matriks Kofaktor dan Adjoin (Transpose)
$$\text{Kofaktor} = \begin{bmatrix} 1 & 2 & -2 \\ -5 & 2 & 1 \\ 1 & -1 & 1 \end{bmatrix} \implies \operatorname{adj}(A) = \begin{bmatrix} 1 & -5 & 1 \\ 2 & 2 & -1 \\ -2 & 1 & 1 \end{bmatrix}$$

#### 4. Menghitung Invers
$$A^{-1} = \frac{1}{\det A} \operatorname{adj}(A) = \frac{1}{3} \begin{bmatrix} 1 & -5 & 1 \\ 2 & 2 & -1 \\ -2 & 1 & 1 \end{bmatrix}$$

**Hasil Akhir:**
$$A^{-1} = \begin{bmatrix} \frac{1}{3} & -\frac{5}{3} & \frac{1}{3} \\ \frac{2}{3} & \frac{2}{3} & -\frac{1}{3} \\ -\frac{2}{3} & \frac{1}{3} & \frac{1}{3} \end{bmatrix}$$

### Soal 8
$$B=\begin{bmatrix}2 & 1 & 0 \\ 1 & 2 & 1 \\ 0 & 1 & 2\end{bmatrix}$$

**Jawaban:**

#### 1. Menghitung Determinan (Ekspansi Baris 1)
$$\det(B) = 2 \begin{vmatrix} 2 & 1 \\ 1 & 2 \end{vmatrix} - 1 \begin{vmatrix} 1 & 1 \\ 0 & 2 \end{vmatrix} + 0 \begin{vmatrix} 1 & 2 \\ 0 & 1 \end{vmatrix}$$
$$\det(B) = 2(4 - 1) - 1(2 - 0) + 0 = 2(3) - 2 = 6 - 2 = 4$$

#### 2. Menghitung Kofaktor
* **Baris 1:**
  $$C_{11} = + \begin{vmatrix} 2 & 1 \\ 1 & 2 \end{vmatrix} = 4 - 1 = 3$$
  $$C_{12} = - \begin{vmatrix} 1 & 1 \\ 0 & 2 \end{vmatrix} = -(2 - 0) = -2$$
  $$C_{13} = + \begin{vmatrix} 1 & 2 \\ 0 & 1 \end{vmatrix} = 1 - 0 = 1$$

* **Baris 2:**
  $$C_{21} = - \begin{vmatrix} 1 & 0 \\ 1 & 2 \end{vmatrix} = -(2 - 0) = -2$$
  $$C_{22} = + \begin{vmatrix} 2 & 0 \\ 0 & 2 \end{vmatrix} = 4 - 0 = 4$$
  $$C_{23} = - \begin{vmatrix} 2 & 1 \\ 0 & 1 \end{vmatrix} = -(2 - 0) = -2$$

* **Baris 3:**
  $$C_{31} = + \begin{vmatrix} 1 & 0 \\ 2 & 1 \end{vmatrix} = 1 - 0 = 1$$
  $$C_{32} = - \begin{vmatrix} 2 & 0 \\ 1 & 1 \end{vmatrix} = -(2 - 0) = -2$$
  $$C_{33} = + \begin{vmatrix} 2 & 1 \\ 1 & 2 \end{vmatrix} = 4 - 1 = 3$$

#### 3. Matriks Kofaktor dan Adjoin (Transpose)
$$\text{Kofaktor} = \begin{bmatrix} 3 & -2 & 1 \\ -2 & 4 & -2 \\ 1 & -2 & 3 \end{bmatrix} \implies \operatorname{adj}(B) = \begin{bmatrix} 3 & -2 & 1 \\ -2 & 4 & -2 \\ 1 & -2 & 3 \end{bmatrix}$$

#### 4. Menghitung Invers
$$B^{-1} = \frac{1}{\det B} \operatorname{adj}(B) = \frac{1}{4} \begin{bmatrix} 3 & -2 & 1 \\ -2 & 4 & -2 \\ 1 & -2 & 3 \end{bmatrix}$$

**Hasil Akhir:**
$$B^{-1} = \begin{bmatrix} \frac{3}{4} & -\frac{1}{2} & \frac{1}{4} \\ -\frac{1}{2} & 1 & -\frac{1}{2} \\ \frac{1}{4} & -\frac{1}{2} & \frac{3}{4} \end{bmatrix}$$

### Soal 9
$$C=\begin{bmatrix}3 & 0 & 2 \\ 2 & 0 & -2 \\ 0 & 1 & 1\end{bmatrix}$$

**Jawaban:**

#### 1. Menghitung Determinan (Ekspansi Baris 1)
$$\det(C) = 3 \begin{vmatrix} 0 & -2 \\ 1 & 1 \end{vmatrix} - 0 + 2 \begin{vmatrix} 2 & 0 \\ 0 & 1 \end{vmatrix}$$
$$\det(C) = 3(0 - (-2)) + 2(2 - 0) = 3(2) + 2(2) = 6 + 4 = 10$$

#### 2. Menghitung Kofaktor
* **Baris 1:**
  $$C_{11} = + \begin{vmatrix} 0 & -2 \\ 1 & 1 \end{vmatrix} = 0 - (-2) = 2$$
  $$C_{12} = - \begin{vmatrix} 2 & -2 \\ 0 & 1 \end{vmatrix} = -(2 - 0) = -2$$
  $$C_{13} = + \begin{vmatrix} 2 & 0 \\ 0 & 1 \end{vmatrix} = 2 - 0 = 2$$

* **Baris 2:**
  $$C_{21} = - \begin{vmatrix} 0 & 2 \\ 1 & 1 \end{vmatrix} = -(0 - 2) = 2$$
  $$C_{22} = + \begin{vmatrix} 3 & 2 \\ 0 & 1 \end{vmatrix} = 3 - 0 = 3$$
  $$C_{23} = - \begin{vmatrix} 3 & 0 \\ 0 & 1 \end{vmatrix} = -(3 - 0) = -3$$

* **Baris 3:**
  $$C_{31} = + \begin{vmatrix} 0 & 2 \\ 0 & -2 \end{vmatrix} = 0 - 0 = 0$$
  $$C_{32} = - \begin{vmatrix} 3 & 2 \\ 2 & -2 \end{vmatrix} = -(-6 - 4) = 10$$
  $$C_{33} = + \begin{vmatrix} 3 & 0 \\ 2 & 0 \end{vmatrix} = 0 - 0 = 0$$

#### 3. Matriks Kofaktor dan Adjoin (Transpose)
$$\text{Kofaktor} = \begin{bmatrix} 2 & -2 & 2 \\ 2 & 3 & -3 \\ 0 & 10 & 0 \end{bmatrix} \implies \operatorname{adj}(C) = \begin{bmatrix} 2 & 2 & 0 \\ -2 & 3 & 10 \\ 2 & -3 & 0 \end{bmatrix}$$

#### 4. Menghitung Invers
$$C^{-1} = \frac{1}{\det C} \operatorname{adj}(C) = \frac{1}{10} \begin{bmatrix} 2 & 2 & 0 \\ -2 & 3 & 10 \\ 2 & -3 & 0 \end{bmatrix}$$

**Hasil Akhir:**
$$C^{-1} = \begin{bmatrix} \frac{1}{5} & \frac{1}{5} & 0 \\ -\frac{1}{5} & \frac{3}{10} & 1 \\ \frac{1}{5} & -\frac{3}{10} & 0 \end{bmatrix}$$
